```python
import json
import csv
import os
```

```python
def save_to_csv(jsonFilePath, csvFilePath, epoch):
    with open(jsonFilePath, 'r') as file:
        data = json.load(file)
    
    num_files = len(list(data.values())[0])

    for i in range(num_files):
        filename = f'trainedACC_{epoch}_{i}.csv'
        file_path = os.path.join(csvFilePath, filename)

        with open(file_path, 'w', newline='') as csvfile:
            csvwriter = csv.writer(csvfile)
            header = ['slope'] + [f'epoch_{i}' for i in range(0,1000)]
            csvwriter.writerow(header)

            for slope, lists in data.items():
                row = [f'{slope}'] + lists[i]
                csvwriter.writerow(row)


def save_to_csv_cor(jsonFilePath, csvFilePath, epoch):
    with open(jsonFilePath, 'r') as file:
        data = json.load(file)

    for slope, matrix in data.items():
        for i in range(len(matrix)):
            fileName = f'cor_output_{epoch}_{slope}_{i}.csv'
            filePath = os.path.join(csvFilePath, fileName)

            with open(filePath, 'w', newline='') as csvfile:
                csvwriter = csv.writer(csvfile)
                header = [f'{slope}'] + [j for j in range(0,50)]
                csvwriter.writerow(header)

                for x in range(0,50):
                    row = [x] + matrix[i][x]
                    csvwriter.writerow(row)


    
        
```

```python

# File paths
epoch = 1000
json_file_path = f'./{epoch}/'  # Update this path to your JSON file

json_file_name_acc = f'trainedACC_{epoch}.json'
csv_file_path_acc = f'./{epoch}/trainedACC_CSV'

json_file_name_cor = f'cor_output_{epoch}_19.json'
csv_file_path_cor = f'./{epoch}/cor_output_CSV'  # Update this path to your desired CSV file location

# Ensure the CSV directory exists
os.makedirs(csv_file_path_acc, exist_ok=True)
os.makedirs(csv_file_path_cor, exist_ok=True)



# Call the function to convert JSON to CSV
save_to_csv(os.path.join(json_file_path, json_file_name_acc), csv_file_path_acc,epoch)
save_to_csv_cor(os.path.join(json_file_path, json_file_name_cor), csv_file_path_cor, epoch)
```
